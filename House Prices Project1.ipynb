## Handling Missing Values

Missing numerical values are replaced using the median, while missing categorical values are replaced using the mode. This improves data quality and prepares the dataset for analysis.

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

missing.sort_values(ascending=False)
```

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print(df.isnull().sum().sum())
```

## Outlier Treatment Using IQR

The Interquartile Range (IQR) method is used to detect and remove extreme values from the SalePrice column, improving the reliability of the dataset.

```python
Q1 = df["SalePrice"].quantile(0.25)

Q3 = df["SalePrice"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

df = df[(df["SalePrice"] >= lower) & (df["SalePrice"] <= upper)]
```

```python
print(df.shape)
```

## Feature Engineering

New features are created from existing variables to improve the predictive power of the dataset. Feature engineering is an important step in data preprocessing.

```python
df["HouseAge"] = df["YrSold"] - df["YearBuilt"]
```

```python
df["TotalBathrooms"] = (
    df["FullBath"] +
    0.5 * df["HalfBath"] +
    df["BsmtFullBath"] +
    0.5 * df["BsmtHalfBath"]
) 
```

```python
df["TotalPorchArea"] = (
    df["OpenPorchSF"] +
    df["EnclosedPorch"] +
    df["3SsnPorch"] +
    df["ScreenPorch"]
)
```

```python
df["TotalLivingArea"] = (
    df["GrLivArea"] +
    df["TotalBsmtSF"]
)
```

### Newly Created Features

- HouseAge: Age of the house when sold.
- TotalBathrooms: Total number of bathrooms.
- TotalPorchArea: Total porch area.
- TotalLivingArea: Total living area.
